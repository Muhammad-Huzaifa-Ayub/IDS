## Scenario 01: University Course Registration System

A university wants to manage course registrations for students. Each student can register for multiple courses, but duplicate registrations must not be allowed.

**Your task:**
- Ask for the student's name and the courses they want to register for.
- Store the registered courses in a **list**.
- Use a **set** to identify and prevent duplicate course registrations.
- Store each course's code and credit hours in a **tuple**.
- Use a **dictionary** to maintain course details, including course codes and credit hours.
- Display the final registered courses, total credit hours, and any duplicate entries rejected.

```python
# Scenario 01: University Course Registration System

name = input("Enter student name: ")

courses = {
    "CS101": ("Introduction to Programming", 3),
    "CS102": ("Computer Fundamentals", 3),
    "MTH101": ("Calculus", 4),
    "ENG101": ("English Composition", 3),
    "PHY101": ("Applied Physics", 3)
}

registered_courses = []
unique_courses = set()
duplicates = []

print("\nAvailable Courses:")
for code, details in courses.items():
    print(code, "-", details[0], "| Credit Hours:", details[1])

while True:
    course_code = input("\nEnter course code or 'done' to finish: ").upper()

    if course_code == "DONE":
        break

    if course_code not in courses:
        print("Invalid course code. Please try again.")
    elif course_code in unique_courses:
        duplicates.append(course_code)
        print("Duplicate registration rejected:", course_code)
    else:
        unique_courses.add(course_code)
        registered_courses.append((course_code, courses[course_code][1]))
        print("Course registered successfully.")

total_credit_hours = 0

print("\n----- Course Registration Summary -----")
print("Student Name:", name)
print("\nRegistered Courses:")

if len(registered_courses) == 0:
    print("No courses registered.")
else:
    for course in registered_courses:
        code = course[0]
        credit_hours = course[1]
        print(code, "-", courses[code][0], "| Credit Hours:", credit_hours)
        total_credit_hours = total_credit_hours + credit_hours

print("\nTotal Registered Courses:", len(registered_courses))
print("Total Credit Hours:", total_credit_hours)

print("\nDuplicate Entries Rejected:")
if len(duplicates) == 0:
    print("No duplicates.")
else:
    for code in duplicates:
        print(code)

print("---------------------------------------")
```

## Scenario 02: Inventory and Stock Management System

A small shop needs a program to monitor its products, prices, and available stock.

**Your task:**
- Allow the shopkeeper to enter product names, prices, and quantities.
- Use a **dictionary** to store product information, with product names as keys.
- Store each product's price and quantity together in a **tuple**.
- Maintain a **set** of unique product categories.
- Use a **list** to record products that need restocking when their quantity falls below five.
- Display the complete inventory, total inventory value, unique categories, and restocking list.
- Allow the shopkeeper to update stock quantities and prices.

```python
# Scenario 02: Inventory and Stock Management System

inventory = {}
categories = set()
restock_list = []

number_of_products = int(input("How many products do you want to enter? "))

while number_of_products < 0:
    print("Number of products cannot be negative.")
    number_of_products = int(input("How many products do you want to enter? "))

for i in range(number_of_products):
    print("\nEnter details for product", i + 1)

    name = input("Product name: ").strip()

    while name == "":
        print("Product name cannot be empty.")
        name = input("Product name: ").strip()

    if name in inventory:
        print("This product already exists. Enter a different product name.")
        continue

    price = float(input("Product price (Rs.): "))
    while price < 0:
        print("Price cannot be negative.")
        price = float(input("Product price (Rs.): "))

    quantity = int(input("Available quantity: "))
    while quantity < 0:
        print("Quantity cannot be negative.")
        quantity = int(input("Available quantity: "))

    category = input("Product category: ").strip()
    while category == "":
        print("Category cannot be empty.")
        category = input("Product category: ").strip()

    inventory[name] = (price, quantity)
    categories.add(category)

    if quantity < 5:
        restock_list.append(name)

while True:
    print("\n----- Inventory Menu -----")
    print("1. Update product price")
    print("2. Update product quantity")
    print("3. Display inventory report")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter product name: ").strip()

        if name in inventory:
            price = float(input("Enter new price (Rs.): "))

            while price < 0:
                print("Price cannot be negative.")
                price = float(input("Enter new price (Rs.): "))

            quantity = inventory[name][1]
            inventory[name] = (price, quantity)
            print("Product price updated successfully.")
        else:
            print("Product not found.")

    elif choice == "2":
        name = input("Enter product name: ").strip()

        if name in inventory:
            quantity = int(input("Enter new stock quantity: "))

            while quantity < 0:
                print("Quantity cannot be negative.")
                quantity = int(input("Enter new stock quantity: "))

            price = inventory[name][0]
            inventory[name] = (price, quantity)
            print("Stock quantity updated successfully.")
        else:
            print("Product not found.")

    elif choice == "3":
        total_value = 0
        restock_list = []

        print("\n----- Complete Inventory -----")

        if len(inventory) == 0:
            print("Inventory is empty.")

        for name, details in inventory.items():
            price = details[0]
            quantity = details[1]
            value = price * quantity
            total_value = total_value + value

            print("Product:", name)
            print("Price: Rs.", format(price, ".2f"))
            print("Quantity:", quantity)
            print("Stock Value: Rs.", format(value, ".2f"))
            print("-----------------------------")

            if quantity < 5:
                restock_list.append(name)

        print("Total Inventory Value: Rs.", format(total_value, ".2f"))
        print("Unique Categories:", categories)

        print("Products Needing Restocking:")
        if len(restock_list) == 0:
            print("No products need restocking.")
        else:
            for product in restock_list:
                print(product)

    elif choice == "4":
        print("Exiting inventory system.")
        break

    else:
        print("Invalid choice. Please select 1, 2, 3, or 4.")
```

## Scenario 03: Event Registration and Attendance Management

A university society organizes an event and needs to manage registrations and attendance.

**Your task:**
- Accept participant names and their registration numbers.
- Use a **dictionary** to associate each registration number with a participant's name.
- Maintain a **set** of registered participants to prevent duplicate registrations.
- Store event details, such as event name, date, and venue, in a **tuple**.
- Use a **list** to record participants who attend the event.
- Display registered participants, attendees, absentees, and total attendance percentage.
- Ensure that unregistered participants cannot be marked as present.

```python
# Scenario 03: Event Registration and Attendance Management

event_name = input("Enter event name: ")
event_date = input("Enter event date: ")
event_venue = input("Enter event venue: ")

event_details = (event_name, event_date, event_venue)

participants = {}
registered_participants = set()
attendees = []

number = int(input("\nHow many participants do you want to register? "))

while number < 0:
    print("Number of participants cannot be negative.")
    number = int(input("How many participants do you want to register? "))

for i in range(number):
    print("\nParticipant", i + 1)

    name = input("Enter participant name: ").strip()
    registration_number = input("Enter registration number: ").strip()

    while name == "" or registration_number == "":
        print("Name and registration number cannot be empty.")
        name = input("Enter participant name: ").strip()
        registration_number = input("Enter registration number: ").strip()

    if registration_number in registered_participants:
        print("Duplicate registration rejected.")
    else:
        participants[registration_number] = name
        registered_participants.add(registration_number)
        print("Participant registered successfully.")

print("\n----- Event Details -----")
print("Event Name:", event_details[0])
print("Date:", event_details[1])
print("Venue:", event_details[2])

print("\n----- Mark Attendance -----")

while True:
    registration_number = input(
        "\nEnter registration number to mark attendance or type 'done': "
    ).strip()

    if registration_number.lower() == "done":
        break

    if registration_number not in registered_participants:
        print("This participant is not registered. Attendance rejected.")
    elif registration_number in attendees:
        print("Attendance has already been marked.")
    else:
        attendees.append(registration_number)
        print("Attendance marked for", participants[registration_number])

absentees = []

for registration_number in registered_participants:
    if registration_number not in attendees:
        absentees.append(registration_number)

total_registered = len(registered_participants)
total_attendees = len(attendees)

if total_registered > 0:
    attendance_percentage = (total_attendees / total_registered) * 100
else:
    attendance_percentage = 0

print("\n===== Event Attendance Report =====")

print("\nRegistered Participants:")
if total_registered == 0:
    print("No participants registered.")
else:
    for registration_number, name in participants.items():
        print(registration_number, "-", name)

print("\nParticipants Present:")
if total_attendees == 0:
    print("No attendees.")
else:
    for registration_number in attendees:
        print(registration_number, "-", participants[registration_number])

print("\nParticipants Absent:")
if len(absentees) == 0:
    print("No absentees.")
else:
    for registration_number in absentees:
        print(registration_number, "-", participants[registration_number])

print("\nTotal Registered:", total_registered)
print("Total Present:", total_attendees)
print("Total Absent:", len(absentees))
print("Attendance Percentage:", format(attendance_percentage, ".2f"), "%")
print("===================================")
```

## Scenario 04: Online Examination Result Analyzer

A teacher wants to analyze examination results for a class of students across multiple subjects.

**Your task:**
- Accept students' names and their marks in multiple subjects.
- Use a **dictionary** to store each student's subject-wise marks.
- Use a **list** to maintain the class roster.
- Use a **tuple** to store the names of subjects included in the examination.
- Use a **set** to identify the distinct grades achieved by the class, using these grade boundaries: A (80–100), B (70–79), C (60–69), D (50–59), and F (below 50).
- Calculate each student's total marks, average marks, and grade.
- Display the class topper, students who failed in at least one subject, and the class average.
- Validate that marks are between 0 and 100.

```python
# Scenario 04: Online Examination Result Analyzer

subjects = ("English", "Mathematics", "Computer Science")

students = []
student_marks = {}
grades = set()

number_of_students = int(input("Enter number of students: "))

while number_of_students <= 0:
    print("Number of students must be greater than zero.")
    number_of_students = int(input("Enter number of students: "))

number_of_subjects = len(subjects)

for i in range(number_of_students):
    name = input("\nEnter student name: ").strip()

    while name == "":
        print("Student name cannot be empty.")
        name = input("Enter student name: ").strip()

    while name in student_marks:
        print("This student name already exists. Enter a different name.")
        name = input("Enter student name: ").strip()

    students.append(name)
    student_marks[name] = {}

    for subject in subjects:
        marks = float(input("Enter marks in " + subject + " (0-100): "))

        while marks < 0 or marks > 100:
            print("Marks must be between 0 and 100.")
            marks = float(input("Enter marks in " + subject + " (0-100): "))

        student_marks[name][subject] = marks

print("\n----- Examination Results -----")

class_total = 0
class_topper = ""
highest_total = -1
failed_students = []

for name in students:
    total = 0
    failed = False

    print("\nStudent:", name)

    for subject in subjects:
        marks = student_marks[name][subject]
        total = total + marks
        print(subject + ":", format(marks, ".2f"))

        if marks < 50:
            failed = True

    average = total / number_of_subjects

    if average >= 80:
        grade = "A"
    elif average >= 70:
        grade = "B"
    elif average >= 60:
        grade = "C"
    elif average >= 50:
        grade = "D"
    else:
        grade = "F"

    grades.add(grade)
    class_total = class_total + total

    print("Total Marks:", format(total, ".2f"))
    print("Average Marks:", format(average, ".2f"))
    print("Grade:", grade)

    if failed:
        failed_students.append(name)

    if total > highest_total:
        highest_total = total
        class_topper = name

class_average = class_total / (number_of_students * number_of_subjects)

print("\n===== Class Result Summary =====")
print("Class Topper:", class_topper)
print("Topper's Total Marks:", format(highest_total, ".2f"))
print("Class Average:", format(class_average, ".2f"))
print("Grades Achieved:", grades)

print("\nStudents Who Failed in at Least One Subject:")
if len(failed_students) == 0:
    print("No students failed in any subject.")
else:
    for name in failed_students:
        print(name)

print("================================")
```

## Scenario 05: Personal Expense Tracker and Spending Analysis

A student wants to track daily expenses and identify where their money is being spent.

**Your task:**
- Accept expense descriptions, categories, and amounts over multiple entries.
- Use a **list** to maintain the expense records.
- Use a **tuple** to represent each expense record, containing its description, category, and amount.
- Use a **dictionary** to calculate total expenditure for each category.
- Use a **set** to display all unique expense categories.
- Allow the student to search for expenses by category.
- Display total expenditure, category-wise spending, the highest individual expense, and the category with the highest total spending.
- Reject negative expense amounts and handle cases where no expenses have been recorded.

```python
# Scenario 05: Personal Expense Tracker and Spending Analysis

expenses = []
category_totals = {}
categories = set()

number_of_expenses = int(input("How many expenses do you want to record? "))

while number_of_expenses < 0:
    print("Number of expenses cannot be negative.")
    number_of_expenses = int(input("How many expenses do you want to record? "))

for i in range(number_of_expenses):
    print("\nExpense", i + 1)

    description = input("Enter expense description: ").strip()
    while description == "":
        print("Description cannot be empty.")
        description = input("Enter expense description: ").strip()

    category = input("Enter expense category: ").strip()
    while category == "":
        print("Category cannot be empty.")
        category = input("Enter expense category: ").strip()

    amount = float(input("Enter expense amount (Rs.): "))

    while amount < 0:
        print("Expense amount cannot be negative.")
        amount = float(input("Enter expense amount (Rs.): "))

    expense_record = (description, category, amount)
    expenses.append(expense_record)

    categories.add(category)

    if category in category_totals:
        category_totals[category] = category_totals[category] + amount
    else:
        category_totals[category] = amount

print("\n----- Expense Summary -----")

if len(expenses) == 0:
    print("No expenses have been recorded.")
else:
    total_expenditure = 0
    highest_expense = expenses[0]
    highest_category = ""
    highest_category_amount = 0

    print("\nAll Expense Records:")

    for expense in expenses:
        description = expense[0]
        category = expense[1]
        amount = expense[2]

        print(description, "|", category, "| Rs.", format(amount, ".2f"))
        total_expenditure = total_expenditure + amount

        if amount > highest_expense[2]:
            highest_expense = expense

    print("\nTotal Expenditure: Rs.", format(total_expenditure, ".2f"))

    print("\nCategory-wise Spending:")
    for category in category_totals:
        print(category, ": Rs.", format(category_totals[category], ".2f"))

        if category_totals[category] > highest_category_amount or highest_category == "":
            highest_category_amount = category_totals[category]
            highest_category = category

    print("\nUnique Expense Categories:", categories)

    print("\nHighest Individual Expense:")
    print("Description:", highest_expense[0])
    print("Category:", highest_expense[1])
    print("Amount: Rs.", format(highest_expense[2], ".2f"))

    print("\nCategory with Highest Total Spending:", highest_category)
    print("Category Total: Rs.", format(highest_category_amount, ".2f"))

    search_category = input("\nEnter a category to search for expenses: ").strip()

    found = False
    print("\nExpenses in Category:", search_category)

    for expense in expenses:
        if expense[1].lower() == search_category.lower():
            print(expense[0], "- Rs.", format(expense[2], ".2f"))
            found = True

    if not found:
        print("No expenses found in this category.")

print("---------------------------")
```
