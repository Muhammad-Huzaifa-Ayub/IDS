## 1. The problem

**Scenario.** The teaching assistant needs a small self-test program for
first-week students. The program should:

1. Welcome the student and ask for their name.
2. Ask a fixed set of multiple-choice questions about Python basics.
3. Accept answers like `A`, `a`, or even ` a ` (real users are messy) and
   be forgiving about case/extra spaces.
4. Tell the student after each question whether they were right.
5. Count the score.
6. At the end, print a **grade report**: name, score, percentage, letter
   grade, pass/fail, and one line of personalised feedback.
7. Optionally run again for the next student.

## 2. Requirements (what "done" means)

| # | Requirement |
|---|---|
| R1 | Reads the student's name as text |
| R2 | Asks exactly 5 multiple-choice questions |
| R3 | Accepts answers in any case, with surrounding spaces ignored |
| R4 | Gives immediate right/wrong feedback per question |
| R5 | Counts correct answers (counter pattern) |
| R6 | Computes percentage and letter grade (A/B/C/D/F) |
| R7 | Prints a formatted report with f-strings |
| R8 | Asks whether another student wants to try (loop) |

## 3. Design — the algorithm in plain words

```text
1. Print a welcome banner.
2. Set running = "yes".
3. While running is "yes":
     a. Read the student's name.
     b. Set score = 0.
     c. For each of the 5 questions:
          - show the question and options,
          - read the student's answer, clean it (strip, upper),
          - if it matches the correct option, add 1 to score
            and print "Correct!", otherwise print the right answer.
     d. Compute percentage = score / 5 * 100.
     e. Decide the letter grade from the percentage.
     f. Print the report.
     g. Ask "Run for another student? (yes/no)" and update running.
4. Print a goodbye line.
```

## 4. Pseudocode

Pseudocode is the algorithm written in structured half-English/half-code —
precise about the *logic*, silent about the *syntax*:

```text
START
  DISPLAY banner
  running <- "YES"
  WHILE running = "YES":
      name <- INPUT "Student name"
      score <- 0
      FOR each (question, options, correct) IN quiz_data:
          DISPLAY question
          DISPLAY options
          answer <- INPUT "Your answer"
          answer <- UPPERCASE( TRIM(answer) )
          IF answer = correct THEN
              score <- score + 1
              DISPLAY "Correct!"
          ELSE
              DISPLAY "Wrong. Correct answer:", correct
          ENDIF
      ENDFOR
      percentage <- score / 5 * 100
      IF percentage >= 80 THEN grade <- "A"
      ELIF percentage >= 60 THEN grade <- "B"
      ELIF percentage >= 40 THEN grade <- "C"
      ELSE grade <- "F"
      ENDIF
      DISPLAY report(name, score, percentage, grade)
      running <- UPPERCASE( TRIM( INPUT "Another student? yes/no" ) )
  ENDWHILE
  DISPLAY "Goodbye"
END
```

Notice how the pseudocode already names the patterns you know: a `WHILE`
sentinel loop around everything, a `FOR` loop over the questions, a counter
`score`, and an `IF/ELIF` chain for the grade.

## 5. The quiz data — one gentle new idea

Five questions, each with its text, its options, and its correct letter.
Python's **tuple** (introduced as "a sequence" in Modules 2 and 6) lets us
group each question's pieces:

````python
question_one = ("Which function displays text on the screen?",
                "A. print()   B. show()   C. display()",
                "A")
question_two = ("What does input() always return?",
                "A. an integer   B. a string   C. a float",
                "B")
question_three = ("Which operator gives the remainder?",
                  "A. /   B. //   C. %",
                  "C")
question_four = ("Which keyword starts a conditional statement?",
                 "A. if   B. for   C. while",
                 "A")
question_five = ("How many times does range(3) loop?",
                 "A. 2   B. 3   C. 4",
                 "B")

# A tuple of five tuples: the whole quiz in one variable.
quiz_data = (question_one, question_two, question_three,
             question_four, question_five)

print("Questions loaded:", len(quiz_data))
````

Each item is used exactly as in Module 6's loops: `for question in
quiz_data:` visits every question in order, and `question[0]`,
`question[1]`, `question[2]` are its text, options and correct letter.

## 6. Implementation — building the program piece by piece

### Step 1 — the banner and the outer loop skeleton

````python
print("=" * 46)
print("PYTHON BASICS SELF-TEST   (5 questions)".center(46))
print("=" * 46)

running = "YES"
print("(Outer loop will run while running == 'YES')")
````

### Step 2 — cleaning answers (Module 4 methods at work)

The single most important robustness trick of the whole project:

````python
raw_answer = "   b  "          # what a real user might type
clean_answer = raw_answer.strip().upper()
print(f"|{clean_answer}|")

is_correct = clean_answer == "B"
print("Correct?", is_correct)
````

`strip()` removes the surrounding spaces, `upper()` makes comparison
case-insensitive (R3 ✓).

### Step 3 — the grading logic (Module 5)

````python
score = 3                       # simulated result: 3 of 5
percentage = score / 5 * 100

if percentage >= 80:
    grade = "A"
elif percentage >= 60:
    grade = "B"
elif percentage >= 40:
    grade = "C"
else:
    grade = "F"

passed = percentage >= 40
print(f"Score {score}/5 = {percentage:.0f}% -> grade {grade}, passed: {passed}")
````

### Step 4 — personalised feedback (conditions on strings and numbers)

````python
# Feedback depends on both the percentage and the name:
percentage = 60.0
student_name = "Amina"

if percentage == 100:
    feedback = "Perfect score — outstanding!"
elif percentage >= 80:
    feedback = f"Excellent work, {student_name}!"
elif percentage >= 60:
    feedback = f"Good effort, {student_name} — review the misses."
elif percentage >= 40:
    feedback = f"You passed, {student_name}, but revisit Modules 2-4."
else:
    feedback = f"Keep practising, {student_name} — try the quiz again."

print(feedback)
````

### Step 5 — one full question round (everything but the loop)

````python
# One round with a simulated typed answer:
# (This cell depends on quiz_data, created several cells above — top-to-bottom
# execution keeps it alive, exactly like a Jupyter session.)
question, options, correct = quiz_data[0]     # unpack the tuple's pieces

print(question)
print(options)
raw_answer = " a "                            # as if the student typed this
clean_answer = raw_answer.strip().upper()

if clean_answer == correct:
    print("Correct!")
else:
    print(f"Wrong — the correct answer was {correct}.")
````

The line `question, options, correct = quiz_data[0]` is **tuple
unpacking**: splitting a tuple into named variables. It is simply a neater
way to write `question = quiz_data[0]` plus two more lines.

## 7. The complete program

Everything assembled, exactly as designed. This cell contains **real
`input()` calls** — when you run it in Jupyter, type into the input boxes:
name first, then five answers (A/B/C), then yes/no. To replay the exact
session shown in the sample run, type: `Amina`, then `a`, `b`, `c`, `a`,
`b`, then `no`.

````python
# ============================================================
#  MINI PROJECT: Python Basics Self-Test with Grade Report
#  Uses only: variables, types, I/O, strings, conditions, loops
# ============================================================
# This cell is fully interactive. The scripted session used to validate it:
# Amina, a, b, c, a, b, no

print("=" * 46)
print("PYTHON BASICS SELF-TEST   (5 questions)".center(46))
print("=" * 46)

running = "YES"

while running == "YES":

    # ----- R1: student name -----
    student_name = input("Student name: ")
    print()
    print(f"Welcome, {student_name}! Answer with A, B or C.")
    print()

    # ----- R5: the score counter starts at zero -----
    score = 0

    # ----- R2: ask every question in order -----
    for question, options, correct in quiz_data:
        print("-" * 46)
        print(question)
        print(options)
        raw_answer = input("Your answer: ")

        # ----- R3: clean the answer -----
        clean_answer = raw_answer.strip().upper()

        # ----- R4: immediate feedback -----
        if clean_answer == correct:
            print("Correct!")
            score += 1
        else:
            print(f"Wrong — the correct answer was {correct}.")

    # ----- R6: percentage and grade -----
    total_questions = len(quiz_data)
    percentage = score / total_questions * 100

    if percentage >= 80:
        grade = "A"
    elif percentage >= 60:
        grade = "B"
    elif percentage >= 40:
        grade = "C"
    else:
        grade = "F"

    passed = percentage >= 40

    # ----- R7: the grade report -----
    print()
    print("=" * 46)
    print("            GRADE REPORT")
    print("=" * 46)
    print(f"  Student    : {student_name}")
    print(f"  Score      : {score} out of {total_questions}")
    print(f"  Percentage : {percentage:.0f}%")
    print(f"  Grade      : {grade}")
    print(f"  Result     : {'PASS' if passed else 'FAIL'}")

    if percentage == 100:
        feedback = "Perfect score — outstanding!"
    elif percentage >= 80:
        feedback = f"Excellent work, {student_name}!"
    elif percentage >= 60:
        feedback = f"Good effort, {student_name} — review the misses."
    elif percentage >= 40:
        feedback = f"You passed, {student_name}, but revisit Modules 2-4."
    else:
        feedback = f"Keep practising, {student_name} — try the quiz again."
    print(f"  Feedback   : {feedback}")
    print("=" * 46)

    # ----- R8: run again? -----
    running = input("Run for another student? (yes/no): ").strip().upper()

print()
print("Thank you for using the self-test. Goodbye!")
````

### How the code maps to the design

| Design step | Code |
|---|---|
| 2. `running` set to "YES" | `running = "YES"` before the `while` |
| 3a. read name | `student_name = input(...)` |
| 3b. score = 0 | `score = 0` — re-created each round |
| 3c. loop over questions | `for question, options, correct in quiz_data:` |
| 3c. clean answer | `raw_answer.strip().upper()` |
| 3c. feedback | `if clean_answer == correct: ... else: ...` |
| 3d. percentage | `score / total_questions * 100` |
| 3e. grade | the `if/elif/else` chain |
| 3f. report | the f-string block |
| 3g. continue? | `running = input(...).strip().upper()` |

**One subtlety worth noticing:** the sentinel loop condition is
`running == "YES"`, and the last line inside the loop *re-reads and
cleans* `running`. If we forgot to update it, the program would loop
forever — the classic `while` mistake from Module 6.

## 8. Testing the program

Good tests try *different kinds* of answers, not just the happy path:

| Test | Input answer | Cleaned | Expected effect |
|---|---|---|---|
| lowercase | `b` | `B` | accepted as B |
| padded | `"  a "` | `A` | accepted as A |
| wrong case+spaces | `" c  "` | `C` | accepted as C |
| invalid letter | `x` | `X` | counted wrong, program continues |
| empty | `` (just Enter) | `` | counted wrong, program continues |
| stop signal | `no` at the prompt | `NO` | program ends politely |

Notice the design choice: invalid answers are simply *wrong*, the quiz
never crashes. Robust programs degrade gracefully.

## 9. Possible improvements

No program is ever finished. In the order you will meet the tools:

1. **Random question order** — shuffle the questions each run (`random`
   module, next course).
2. **More questions** — load them from a *file* (file I/O course) instead
   of hard-coding.
3. **Functions** — wrap question-asking, grading and report-printing into
   reusable functions (functions course). You will notice the program
   already has clean "chunks" that want to become functions.
4. **Lists and dictionaries** — store many quizzes, track per-question
   results, build a class leaderboard.
5. **Real input validation** — reject invalid letters with a re-ask loop
   (`while clean_answer not in ("A", "B", "C"): ...`).
6. **Timing** — measure how long each question took (`time` module).
7. **A GUI or web version** — once you know classes and frameworks.

Each improvement is small *now* that the structure is clean. That is the
deepest lesson of this project: **programs are designed, then built, then
improved — in that order.**
